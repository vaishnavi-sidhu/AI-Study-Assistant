```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

INPUT_IMAGE = "/content/Screenshot 2026-07-23 113126.png"

img = cv2.imread(INPUT_IMAGE)

if img is None:
    raise FileNotFoundError(
        "Image load nahi hui. Please check that the image is uploaded."
    )

print("✅ Image successfully loaded!")
print("Image dimensions:", img.shape)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

height, width = img.shape[:2]
channels = img.shape[2]

print("\n========== IMAGE PROPERTIES ==========")
print("Width:", width)
print("Height:", height)
print("Dimensions:", img.shape)
print("Number of Channels:", channels)
print("Data Type:", img.dtype)
print("=======================================")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,5))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

print("\n✅ RGB/BGR → Grayscale")
print("✅ BGR → HSV")
print("✅ BGR → LAB")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

resized = cv2.resize(img, (300, 300))

print("\n✅ Image resized to 300 x 300")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
print("✅ Image rotated 90 degrees clockwise")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

horizontal_flip = cv2.flip(img, 1)
vertical_flip = cv2.flip(img, 0)

print("✅ Horizontal flip completed")
print("✅ Vertical flip completed")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

negative = 255 - img

print("✅ Negative image created")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

roi_height = height // 2
roi_width = width // 2

start_y = height // 4
start_x = width // 4

roi = img[
    start_y:start_y + roi_height,
    start_x:start_x + roi_width
]

print("✅ ROI created")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

cv2.imwrite("/content/grayscale.jpg", gray)
cv2.imwrite("/content/resized.jpg", resized)
cv2.imwrite("/content/rotated.jpg", rotated)
cv2.imwrite("/content/horizontal_flip.jpg", horizontal_flip)
cv2.imwrite("/content/vertical_flip.jpg", vertical_flip)
cv2.imwrite("/content/negative.jpg", negative)
cv2.imwrite("/content/roi.jpg", roi)

print("\n✅ All processed images saved successfully!")
```
